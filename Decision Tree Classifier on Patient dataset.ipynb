```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv(r"https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv")
```

```python
df.head()
```

```python
x = df.drop("Outcome",axis=1)
y = df["Outcome"]
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
CT = DecisionTreeClassifier(criterion='gini',max_depth=3)
CT.fit(x_train,y_train)
```

```python
y_pred = CT.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.tree import plot_tree
```

```python
CT.score(x_test,y_test)
```

```python
CT.score(x_train,y_train)
```

```python
plt.figure(figsize=(17,10))
plot_tree(CT,feature_names = x.columns,class_names = ['Diabetic','Non-Diabetic'],filled = True,rounded = True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("screentime.csv")
```

```python
dfn = df.drop("user_id",axis=1)
```

```python
dfn.info()
```

```python
target = {"Optimal Recovery":1,"Mild Deficit":2,"Moderate Debt":3,"Severe Sleep Debt":4}
```

```python
dfn["Target values"] = dfn["sleep_debt_category"].map(target)
```

```python
dfn2 = dfn.drop("sleep_debt_category",axis=1)
```

```python
dfn3 = pd.get_dummies(dfn2,dtype = int)
```

```python
x = dfn3.drop("Target values",axis=1)
y = dfn3["Target values"]
```

```python
x.shape
```

```python
y.shape
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
Logistic = LogisticRegression()
Logistic.fit(x_train,y_train)
```

```python
y_pred = Logistic.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
accuracy_score(y_pred,y_test)
```

```python
print(classification_report(y_pred,y_test))
```

```python
print(confusion_matrix(y_pred,y_test))
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
CT = DecisionTreeClassifier()
CT.fit(x_train,y_train)
CT_PRED = CT.predict(x_test)
```

```python
print(classification_report(y_test,CT_PRED))
```

```python
print(confusion_matrix(y_test,CT_PRED))
```
